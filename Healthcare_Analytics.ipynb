```python
!pip install pandas scikit-learn

import pandas as pd

from sklearn.tree import DecisionTreeClassifier

data=pd.DataFrame({
"Age":[22,35,60,45],
"BP":[110,140,180,150],
"Disease":[0,0,1,1]
})

X=data[
["Age","BP"]
]

y=data[
"Disease"
]

model=DecisionTreeClassifier()

model.fit(
X,
y
)

print(
model.predict(
[[50,160]]
)
)
```
