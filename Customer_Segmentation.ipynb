```python
!pip install scikit-learn

import pandas as pd

from sklearn.cluster import KMeans

data=pd.DataFrame({
"Income":[20,25,80,90],
"Spending":[30,40,85,95]
})

model=KMeans(
n_clusters=2
)

data["Cluster"]=model.fit_predict(
data
)

print(data)
```
