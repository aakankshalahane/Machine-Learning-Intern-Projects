```python
!pip install pandas scikit-learn

import pandas as pd

from sklearn.linear_model import LinearRegression

data=pd.DataFrame({
"Month":[1,2,3,4,5],
"Sales":[100,120,130,160,180]
})

X=data[["Month"]]

y=data["Sales"]

model=LinearRegression()

model.fit(X,y)

print(
model.predict([[6]])
)
```
